# Extension Step 4 — GNN DAGNN Task Classifier v2 (Difference Fusion)

Trains a `DAGClassifier` with **difference fusion** on task graph realizations.

### Architecture change vs v1 (gnn_task_classifier.ipynb)

| Component | v1 (concat) | v2 (difference) |
|---|---|---|
| `NodeFusionProjector` | `concat([t, v]) → Linear(512→256)` | `(t − v) → Linear(256→256)` |
| `hidden_dim` | 128 | 32 |
| `num_layers` | 2 | 1 |
| Approx params | ~328K | ~80K |

**Semantic motivation of `t - v`**: in the EgoVLP joint space, `t - v` measures the gap
between the expected step description (`t`) and the observed visual execution (`v`).
Correct execution: `t ≈ v → t - v ≈ 0`. Error: `t - v` points in a characteristic direction.
Unmatched nodes (`v = 0`): `t - 0 = t`, so the text description is preserved as projector input.

**Prerequisites:**
- Step embeddings in `STEP_EMBEDDINGS_DIR` (output of Extension Step 1)
- Pre-fusion cache in `CACHE_DIR` (same cache as v1 — no rebuild needed)
- Task graphs in `GRAPHS_DIR` (annotations submodule)
- EgoVLP checkpoint at `EGOVLP_CKPT`

**Output:** checkpoints + `results.csv` in `STEP4_OUTPUT_DIR`

## Full training — 50 epochs

Allena il `DAGClassifier` v2 con lo split scelto in `SPLIT_MODE`.

### Configurazioni supportate

| `SPLIT_MODE` | Protocollo | Note |
|---|---|---|
| `"loo"` | LOO 24-fold (same as B2) | confronto diretto con TaskVerifier |
| `"official"` | 213/62/109 recordings | split ufficiale CaptainCook4D |
| `"recipe"` | 16/4/4 recipes, seed=42 | split per ricetta |

### Ablation baseline

Per osservare empiricamente l'overfitting ancora presente con d_h=128 L=2 (difference fusion):
cambia `--hidden_dim 128 --num_layers 2` nel comando sotto e usa un output dir separato.

**Tempo stimato:** ~15-30 min su T4 per official/recipe; ~6-10 ore per LOO.

```python
import subprocess, sys, os
os.chdir('/content/code')

common = [
    '--annotations_path',    ANNOTATIONS_PATH,
    '--step_embeddings_dir', LOCAL_EMBEDDINGS_DIR,
    '--graphs_dir',          GRAPHS_DIR,
    '--egovlp_repo',         EGOVLP_REPO,
    '--egovlp_ckpt',         EGOVLP_CKPT,
    '--cache_dir',           CACHE_DIR,
    '--output_dir',          STEP4_OUTPUT_DIR,
    '--num_epochs',  '50',
    '--hidden_dim',  '32',   # v2 default — change to 128 for ablation baseline
    '--num_layers',  '1',    # v2 default — change to 2 for ablation baseline
    '--dropout',     '0.5',
    '--lr',          '1e-3',
    '--weight_decay','1e-4',
    '--batch_size',  '4',
    '--threshold',   '0.5',
    '--seed',        '42',
    '--num_workers', '2',
    '--enable_wandb',
]

if SPLIT_MODE == 'loo':
    cmd = [sys.executable, '-m', 'extension.step4.train_dag_classifier',
           *common, '--split_mode', 'loo']
elif SPLIT_MODE == 'recipe':
    cmd = [sys.executable, '-m', 'extension.step4.train_dag_classifier',
           *common, '--split_mode', 'recipe', '--recipe_split_seed', '42']
else:  # official
    cmd = [sys.executable, '-m', 'extension.step4.train_dag_classifier',
           *common, '--split_mode', 'official', '--splits_json', SPLITS_JSON]

subprocess.run(cmd, check=True)
```
